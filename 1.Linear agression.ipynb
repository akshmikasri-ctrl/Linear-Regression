#  Step 4  Data Preparation

* Filter input and output Data.

```python
cars_data
```

```python
x = cars_data[["HP","VOL","SP","WT"]]
x
```

```python
y = cars_data["MPG"]
y
```

# Step 5 Model Building

```python
Linear_Model = LinearRegression()
```

# Test 6 Model Training

```python
Linear_Model.fit(x,y)
```

```python
Linear_Model.intercept_
```

```python
Linear_Model.coef_
```

# Test 7 Model Testing

```python
y_predicted = Linear_Model.predict(x)
y_predicted
```

# Test 8 Model Evalution

```python
error = y -  y_predicted
error
```

### let's now perform Homoscadasticity Test from Data preparation
* This test will be performed only after the error

#### It tests the varience of the error should be same  or not

```python
sns.scatterplot(data=cars_data,x="HP",y=error)
plt.show()
```
